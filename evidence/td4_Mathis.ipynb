## 1. Création de la branche de fonctionnalité

Depuis `main`, nous avons créé une branche dédiée à l’affichage des dates :

```bash
git switch -c feature/display-observation-dates
```

L’option `-c` crée la branche et nous positionne dessus.

Pour vérifier la branche active :

```bash
git branch --show-current
```

Résultat :

```text
feature/display-observation-dates
```

Les modifications ont été réalisées sur cette branche, dans `src/main.py`. Le nom du fichier `main.py` est indépendant du nom de la branche Git.

## 2. Ajout des dates dans le programme

Sur la branche `feature/display-observation-dates`, nous avons ajouté deux fonctions dans `src/main.py` :

```python
def get_first_date(prices):
    return prices[0]["date"]


def get_last_date(prices):
    return prices[-1]["date"]
```

- `prices[0]` récupère la première observation.
- `prices[-1]` récupère la dernière observation.
- `["date"]` récupère la date correspondante.

Ces fonctions supposent que la liste est non vide et classée par date croissante.

Dans `display_market_summary`, nous avons ajouté :

```python
print(f"First date  : {get_first_date(prices)}")
print(f"Last date   : {get_last_date(prices)}")
```

Cette fonction étant utilisée pour l’instrument et le benchmark, les dates sont affichées pour les deux.M

## 3. Exécution et vérification

Nous avons exécuté le programme depuis le terminal :

```bash
python src/main.py
```

Les premières et dernières dates sont correctement affichées pour Apple et le S&P 500. Le nombre d’observations et les prix de clôture restent également présents.

## 4. Préparation et création du commit

Nous avons vérifié l’état du dépôt :

```bash
git status
```

Git indiquait que `src/main.py` avait été modifié, mais n’était pas encore préparé pour un commit.

Nous avons inspecté les modifications, puis préparé le fichier :

```bash
git diff
git add src/main.py
git --no-pager diff --staged
```

- `git diff` affiche les modifications non préparées.
- `git add` prépare les changements pour le prochain commit.
- `git diff --staged` affiche les changements préparés.
- `--no-pager` affiche le résultat directement dans le terminal.

Après vérification, nous avons créé le commit :

```bash
git commit -m "feat: display first and last observation dates"
```

Puis nous avons vérifié :

```bash
git status
git log --oneline -3
```

Résultats :

```text
On branch feature/display-observation-dates
nothing to commit, working tree clean
```

```text
b955bee (HEAD -> feature/display-observation-dates) feat: display first and last observation dates
9596943 (origin/main, origin/HEAD, main) Add files via upload
f24b71c docs: add simple comments to MarketPulse code
```

Le commit `b955bee` enregistre la fonctionnalité sur sa branche. À cette étape, `main` reste sur `9596943` : les modifications ne sont pas encore intégrées à `main`.

`working tree clean` signifie qu’aucune modification locale n’est en attente d’enregistrement.

## 5. Comparaison avec la branche main

Avant la fusion, nous avons consulté l’historique des branches :

```bash
git --no-pager log --oneline --graph --decorate --all
```

Cette commande affiche les commits sous forme de graphe et indique la position des branches.

À cette étape :

| Branche | Commit |
|---|---|
| `feature/display-observation-dates` | `b955bee` |
| `main` | `9596943` |

Nous avons ensuite comparé le contenu des deux branches :

```bash
git --no-pager diff main..feature/display-observation-dates
```

Extraits du résultat :

```diff
+def get_first_date(prices):
+    return prices[0]["date"]
+
+def get_last_date(prices):
+    return prices[-1]["date"]
```

```diff
+    print(f"First date  : {get_first_date(prices)}")
+    print(f"Last date   : {get_last_date(prices)}")
```

Les lignes précédées de `+` correspondent aux ajouts par rapport à `main`.

La comparaison confirme que les changements concernent uniquement `src/main.py` et que la fonctionnalité est encore isolée sur sa branche.

## 6. Fusion dans main

Après vérification, nous sommes revenus sur `main` :

```bash
git switch main
```

Sortie obtenue :

```text
Switched to branch 'main'
Your branch is up to date with 'origin/main'.
```

Nous avons ensuite intégré la fonctionnalité :

```bash
git merge feature/display-observation-dates
```

Sortie obtenue :

```text
Updating 9596943..b955bee
Fast-forward
 src/main.py | 9 +++++++++
 1 file changed, 9 insertions(+)
```

### Comprendre le fast-forward

`main` n’avait pas reçu de nouveau commit depuis la création de la branche de fonctionnalité. Git a donc simplement avancé `main` de `9596943` à `b955bee`.

Aucun commit de fusion supplémentaire n’a été créé et aucun conflit n’est apparu.

La branche `feature/display-observation-dates` existe toujours : fusionner une branche ne la supprime pas.

### Vérification après fusion

Depuis `main`, nous avons relancé :

```bash
python src/main.py
```

Le programme affichait correctement les dates pour l’instrument et le benchmark. La fonctionnalité était donc intégrée et fonctionnelle sur `main`.

Cette fusion était locale : la commande `git merge` ne publie pas les changements sur GitHub.M

## 8. Comprendre les conflits

Aucun conflit n’est apparu pendant notre fusion.

Un conflit peut se produire lorsque deux branches modifient différemment les mêmes lignes et que Git ne peut pas les combiner automatiquement.

Pour le résoudre, il faut examiner les versions concernées, choisir le contenu final et supprimer les marqueurs de conflit. On prépare ensuite les fichiers corrigés avec `git add`, puis on termine la fusion avec `git commit`.
