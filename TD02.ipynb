```python
import csv
def load_prices():
    with open("data/sample/prices.csv", encoding="utf-8") as file:
        return list(csv.DictReader(file))
```

```python
def filter_prices(prices, ticker):
    return [row for row in prices if row["ticker"] == ticker]
```

```python
instrument_prices = filter_prices(
    prices,
    instrument["ticker"],
)

benchmark_prices = filter_prices(
    prices,
    benchmark["ticker"],
)
```

```python
print(len(instrument_prices))
print(len(benchmark_prices))
```

```python
def get_first_close(prices):
    first_close_AAPL = prices[0]["close"]
    first_close_SP500 = prices[1]["close"]
    return float(first_close_AAPL), float(first_close_SP500)

get_first_close(prices)
```

```python
def get_last_close(prices):
    last_close_AAPL = prices[-1]["close"]
    last_close_SP500 = prices[-2]["close"]
    return float(last_close_AAPL), float(last_close_SP500)

get_last_close(prices)
```

```python
def display_market_summary(asset, prices, show_currency=True):
    rows = [row for row in prices if row["ticker"] == asset["ticker"]]

    print(asset["ticker"], asset["name"])
    print(len(rows))
    print(rows[0]["close"])
    print(rows[-1]["close"])

    if show_currency:
        print(asset["currency"])
```
